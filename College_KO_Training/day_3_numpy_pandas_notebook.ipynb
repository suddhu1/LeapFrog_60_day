## Section 1 — NumPy 

NumPy (**Num**erical **Py**thon) is the backbone of scientific computing in Python.  
Almost every data library (Pandas, scikit-learn, TensorFlow…) builds on top of it.

### Why NumPy?
Python lists are flexible — they can hold any mix of data types. But that flexibility comes at a cost. When you do arithmetic on a list, Python processes each element one at a time, in a loop, checking the type of each item.

NumPy arrays store **one data type only**, which allows them to be stored as a contiguous block of memory. Operations run in compiled C code, not interpreted Python. The result is speed improvements of **10x to 100x** compared to equivalent Python loops.

For a dataset with 1 million numbers, the difference between a NumPy operation and a Python loop can be the difference between milliseconds and minutes.

Beyond speed, NumPy gives you:
- Multi-dimensional arrays (matrices, tensors)
- Broadcasting — operations between arrays of different shapes
- A complete library of mathematical and statistical functions
- The foundation that Pandas DataFrames are built on

### Creating Arrays

### Vectorised Operations

Operations apply **element-wise** — no `for` loop needed.  
A scalar is automatically broadcast to match the array's shape.

One of the most important skills in NumPy is recognising when you can replace a Python loop with a vectorized operation. The result is the same but dramatically faster.

The key mindset shift: **think in terms of whole arrays, not individual elements.**

```python
# Slow — Python loop
result = []
for x in data:
    result.append(x * 2 + 1)

# Fast — vectorized
result = data * 2 + 1
```


````python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

# Vectorized operations
print("Array Addition:", a + b)
print("Array Subtraction:", a - b)
print("Array Multiplication:", a * b)
print("Array Division:", a / b) 
````

### The `axis` Parameter

For 2D arrays the `axis` keyword controls the *direction* of the operation:

```
axis=0  →  operate DOWN the rows  →  one result per COLUMN
axis=1  →  operate ACROSS columns →  one result per ROW
```

A useful memory trick: `axis=0` **collapses rows** (the 0th dimension); `axis=1` **collapses columns**.

## Reshaping

Reshaping changes the dimensions of an array without changing its data. The total number of elements must remain the same.

| Method | What it does |
|--------|-------------|
| `.reshape(shape)` | Return array with new shape |
| `.flatten()` | Return a 1D copy of the array |
| `.ravel()` | Return a 1D view (no copy) |
| `.T` | Transpose — swap rows and columns |

You can use `-1` in reshape as a wildcard — NumPy calculates that dimension automatically. For example, `arr.reshape(2, -1)` means 2 rows, figure out the columns yourself.

````python
import numpy as np
flat = np.arange(12)                 # [0  1  2 … 11]
print("Original shape :", flat.shape)

# Reshape
print("Reshaped to 3x4 :", flat.reshape(3, 4))

# Flatten

print("Flattened back to 1D :", flat.reshape(-1))
# Transpose the matrix (swap rows and columns)
print("Transposed (swap rows and columns) :\n", flat.reshape(3, 4).T)
````

## Section 2 — Pandas

NumPy gave us powerful numerical arrays. But real-world data is rarely just numbers in a grid — it has column names, row labels, mixed data types, and missing values. **Pandas** is built exactly for this kind of data.

Pandas is the most widely used data analysis library in Python. It is built on top of NumPy, so everything you learned about NumPy carries forward — Pandas adds a richer, labelled structure on top.

Pandas has two core data structures:

| Structure | What it is | Equivalent to |
|-----------|-----------|---------------|
| `Series` | 1-dimensional labelled array | A single column of data |
| `DataFrame` | 2-dimensional labelled table | A full spreadsheet or SQL table |



## Why Pandas?

With a Python list or NumPy array, data has no labels — you access it by position. If you have student data in an array, you have to remember that column 0 is name, column 1 is age, column 2 is GPA. This is fragile and confusing.

Pandas assigns **labels** to both rows and columns. You access data by name, not by position. This makes your code readable, your analysis clear, and your errors much easier to catch.

Beyond labels, Pandas gives you:
- Built-in handling of missing values
- Tools to load data from CSV, Excel, JSON, and databases
- Powerful grouping, filtering, and aggregation
- Integration with Matplotlib for visualisation
- The ability to work with datasets of millions of rows efficiently

By convention, Pandas is always imported as `pd`:

## Pandas Series

A **Series** is a one-dimensional array with an **index** — a label attached to each value. If you think of a dictionary, a Series is like a more powerful dictionary that supports arithmetic, statistics, and alignment.

```
Index    Value
  0   →   88
  1   →   76
  2   →   91
  3   →   55
```

The index does not have to be numbers — it can be strings (like student names), dates, or anything unique.

### Creating a Series

```python
pd.Series(data, index=index_labels)
```

You can create a Series from:
- A Python list
- A Python dictionary — keys become the index
- A NumPy array
- A single scalar value


### Series Attributes

| Attribute | What it returns |
|-----------|----------------|
| `.values` | The underlying NumPy array of values |
| `.index` | The index labels |
| `.dtype` | Data type of the values |
| `.shape` | Shape as a tuple |
| `.size` | Total number of elements |
| `.name` | Name of the Series |

````python
import pandas as pd

# Create a Series from a list with default numeric index
scores = [85, 90, 78, 92, 88]
scores_series = pd.Series(scores)
print(scores_series)

# Create a Series from a list with custom string index (student names)
student_index = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
scores_series_custom = pd.Series(scores, index=student_index)
print(scores_series_custom)

# Create a Series from a dictionary — keys become the index automatically
scores_dict = {'Alice': 85, 'Bob': 90, 'Charlie': 78, 'David': 92, 'Eve': 88}
scores_series_dict = pd.Series(scores_dict)
print(scores_series_dict)

````

````python
import pandas as pd

scores = pd.Series(
    [88, 76, 91, 55, 83],
    index=["Aarav", "Sita", "Bikash", "Nisha", "Rohan"],
    name="Exam Scores"
)

print(scores)
# Print: values, index, dtype, shape, size, name
print("Values:", scores.values)
print("Index:", scores.index)
print("Data Type:", scores.dtype)
print("Shape:", scores.shape)
print("Size:", scores.size)
print("Name:", scores.name)

````


### Selecting Columns

`df[['col1', 'col2']]` → **DataFrame** (multiple columns — note the double brackets)

This distinction matters: some operations only work on DataFrames, others only on Series.

### Selecting Rows — `.loc[]` and `.iloc[]`

| | Selector | Use |
|---|---|---|
| `.loc[label]` | **Label-based** | Row label / column name |
| `.iloc[position]` | **Position-based** | Integer index (0, 1, 2 …) |

Both accept slices and lists.  
`.iloc` slicing is **exclusive** of the stop index (Python convention).  
`.loc` slicing is **inclusive** of the stop label.


````python
# First 3 rows by position
print(df.iloc[:3])


# Specific rows by position
print(df.iloc[[0, 3, 5]])

# loc — row label + column name
print(df.loc[0, 'name'])  # 0 le first row ani name le column name wala dincha 

# loc slice (inclusive on both ends)
print(df.loc[1:4, 'name':'owner'])  # rows 1 to 4, columns name to owner (inclusive)
````
